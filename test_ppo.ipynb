# Lab 5 : MiniGrid Dynamic Obstacle 8x8

Test notebook for `main.py`: **PPO** on **MiniGrid-Dynamic-Obstacles-8x8-v0**.

```python
from main import ENV_NAME, SEED, train, evaluate, record_video
import gymnasium as gym
from minigrid.wrappers import ImgObsWrapper
```

## Environment test

```python
env = gym.make(ENV_NAME, render_mode="rgb_array")
env = ImgObsWrapper(env)
obs, info = env.reset(seed=SEED)

print(f"Observation space: {env.observation_space}")
print(f"Action space: {env.action_space}")
print(f"Observation shape: {obs.shape}, dtype: {obs.dtype}")
print(f"Reset info: {info}")
env.close()
```

## 1. `train()`

Trains PPO with a CNN feature extractor for 200,000 timesteps and writes `model.zip` plus TensorBoard logs.

```python
model = train()
```

## Evaluate

```python
mean_reward, std_reward = evaluate(model, n_episodes=10)
print(f"Returned: mean={mean_reward:.2f}, std={std_reward:.2f}")
```

## Record Video

```python
record_video(model, filename="demo.mp4")
```
